```python
# import libraries
import time
from google import genai
import csv
import pandas as pd
import random

import matplotlib.pyplot as plt
```

```python
# get api key
client = genai.Client(api_key="Insert Gemini API Key")
```

```python
# load event titles and descriptions from a CSV file
# if you're running this in Google Colab, upload the CSV first
# from google.colab import files
# files.upload()   # pick SHUevents.csv from your computer

events_csv_path = "SHUevents.csv"
df = pd.read_csv(events_csv_path)
df.head()
```

```python
# passes a random title from the spreadsheet into a prompt that asks gemini to generate an event title based on the given title prompt

i = 0
j = 0
ai_titles = []
ai_descriptions = []

def getDescription(aiTitle):
  # passes each ai-generated title as a prompt to create an ai-generated description
  # using @@@ as the prefix and suffix for the response to ease of reference
  desPrompt = f"Given this University Event Title {aiTitle} Create a brief 10-word Event Description for College students. Place @@@ as a prefix and a suffix before your generated problem."
  desResponse = client.models.generate_content(
    model="gemini-2.0-flash",
    contents=desPrompt,
  )
  time.sleep(1) # adds a 1-second delay between requests to avoid server crash
  return desResponse.text


for i in range(3):      # change range parameter from 3 to len(df)
  j = random.randint(0, len(df)-1)
  title = df['Title'][j]
  #print(title) #check

  # passes each title cell from the spreadsheet as a prompt to create an ai-generated title
  # using ### as the prefix and suffix for the response to ease of reference
  titlePrompt = f"Based on this University Event Title {title} Create one new Event Title for College students. Place ### as a prefix and a suffix before your generated problem."
  titleResponse = client.models.generate_content(
    model="gemini-2.0-flash",
    contents=titlePrompt,
  )
  ai_titles.append(titleResponse.text)
  ai_descriptions.append(getDescription(titleResponse.text))

  #print(titleResponse.text, getDescription(titleResponse.text), '\n') #check
  time.sleep(3) # adds a 3-second delay between requests to avoid server crash

# print list of ai-generated titles and descriptions
for i in range(3):      # change range parameter from 3 to len(df)
  print(ai_titles[i], ai_descriptions[i], '\n')
```

**Key:**

H = human-written event

A = ai-generated event

**Events:**

event 1: H (columns 2-6)

event 2: A (columns 7-11)

event 3: A (columns 12-16)

event 4: H (columns 17-21)

event 5: A (columns 22-26)

event 6: H (columns 27-31)

event 7: H (columns 32-36)

event 8: A (columns 37-41)

event 9: H (columns 42-46)

event 10: A (columns 47-51)

algorithm:

1. Sort df data into two lists: AI-generated event feedback and Human-written event feedback

2. Sort list data into 5 subcategories by each individual event (5 sublists/events per 2 parent lists/categories)

3. Reorganize sublist data into 5 sublists by each parameter (5 parameters for 5 sublists)

5 sub-dataframes of question parameters for ai-generated events

```python
# displays parameter 1 for all ai-generated events
#aiP1_df.head()
```

```python
# new sub df for ai-generated parameter 1
aiP1_df = ai_df.iloc[:, [0, 5, 10, 15, 20]]
#aiP1_df.head()

def sort_by_value(df):
    # create 5 lists for values 1-5
    lists = {1: [], 2: [], 3: [], 4: [], 5: []}
    # iterate through all values in the dataframe
    for value in df.values.flatten():
        lists[value].append(value)
    return lists[1], lists[2], lists[3], lists[4], lists[5]
ones, twos, threes, fours, fives = sort_by_value(aiP1_df)

# calculate frequencies (length of each list)
frequencies = [len(ones), len(twos), len(threes), len(fours), len(fives)]
labels = ['1s', '2s', '3s', '4s', '5s']

# create pie chart
plt.figure(figsize=(4, 4))
plt.pie(frequencies, labels=labels, autopct='%1.1f%%', startangle=90,
        colors=['#FFB6C1', '#FFC0CB', '#FF69B4', '#FF1493', '#C71585'],
        wedgeprops={'edgecolor': 'black'})
plt.title('Appeal of AI-Generated Events', fontsize=14)
plt.axis('equal')
plt.show()


print(len(ones), len(twos), len(threes), len(fours), len(fives))
```
